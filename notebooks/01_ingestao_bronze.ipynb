```python
# TODO F2: Dicionário de Dados (variável, tipo, descrição, exemplo, % nulos)
import pandas as pd

descricoes = {
    'data': 'Data do registro do sinistro',
    'hora': 'Horário registrado do sinistro',
    'bairro': 'Bairro onde ocorreu o sinistro',
    'endereco': 'Logradouro onde ocorreu o sinistro',
    'numero': 'Número do endereço informado',
    'situacao': 'Situação registrada da ocorrência',
    'tipo': 'Tipo de sinistro registrado',
    'natureza': 'Natureza do sinistro',
    'natureza_acidente': 'Natureza do acidente',
    'vitimas': 'Quantidade de vítimas registrada',
    'vitimasfatais': 'Quantidade de vítimas fatais registrada',
    'tempo_clima': 'Condição climática registrada',
    'condicao_via': 'Condição da via registrada',
    'auto': 'Quantidade de automóveis envolvidos',
    'moto': 'Quantidade de motocicletas envolvidas',
    'ciclista': 'Quantidade de ciclistas envolvidos',
    'pedestre': 'Quantidade de pedestres envolvidos',
    'onibus': 'Quantidade de ônibus envolvidos',
    'caminhao': 'Quantidade de caminhões envolvidos',
    'Protocolo': 'Identificador de protocolo da ocorrência',
    'DATA': 'Data do registro do sinistro',
'acidente_verificado': 'Localização do sinistro em relação à via ou cruzamento',
'bairro_cruzamento': 'Bairro associado ao cruzamento informado',
'ciclom': 'Quantidade registrada de ciclomotores envolvidos',
'complemento': 'Informações complementares sobre o local do sinistro',
'conservacao_via': 'Estado de conservação da via',
'descricao': 'Descrição textual da ocorrência',
'detalhe_endereco_acidente': 'Detalhamento do endereço do sinistro',
'divisao_via1': 'Primeira característica registrada da divisão da via',
'divisao_via2': 'Segunda característica registrada da divisão da via',
'divisao_via3': 'Terceira característica registrada da divisão da via',
'endereco_cruzamento': 'Logradouro do cruzamento associado ao sinistro',
'mao_direcao': 'Sentido de circulação permitido na via, como mão única ou dupla',
'num_semaforo': 'Número de identificação do semáforo',
'numero_cruzamento': 'Número do endereço associado ao cruzamento',
'outros': 'Quantidade registrada de outros tipos de veículos envolvidos',
'ponto_controle': 'Tipo de controle de tráfego presente no local',
'referencia_cruzamento': 'Ponto de referência associado ao cruzamento',
'sentido_via': 'Sentido de deslocamento registrado na via',
'sinalizacao': 'Condição da sinalização viária',
'situacao_placa': 'Situação das placas de sinalização no local',
'situacao_semaforo': 'Condição de funcionamento do semáforo',
'velocidade_max_via': 'Velocidade máxima permitida na via',
'viatura': 'Quantidade registrada de viaturas envolvidas'
}

# ATENÇÃO (auditoria F1): o significado de "tipo" e "natureza" muda conforme o ano.
# 2015: tipo = com/sem vítima, natureza = tipo de colisão
# 2016–2021: natureza_acidente = com/sem vítima, tipo = tipo de colisão
# 2022–2024: natureza = com/sem vítima, tipo = tipo de colisão
# Por isso a descrição dessas colunas é decidida pelo CONTEÚDO de cada arquivo.
DESC_CLASSIFICACAO = 'Classificação do sinistro quanto a vítimas (SEM VÍTIMA, COM VÍTIMA, VÍTIMA FATAL)'
DESC_TIPO_COLISAO = 'Tipo de sinistro (colisão, colisão traseira, atropelamento etc.)'

def descrever(coluna, serie):
    if coluna in ('tipo', 'natureza', 'natureza_acidente'):
        eh_classificacao = serie.dropna().str.upper().str.contains('VÍTIMA|VITIMA').mean() > 0.5
        return DESC_CLASSIFICACAO if eh_classificacao else DESC_TIPO_COLISAO
    return descricoes.get(coluna, 'Descrição a validar')

NUMERICAS = {'auto', 'moto', 'ciclom', 'ciclista', 'pedestre', 'onibus', 'caminhao',
             'viatura', 'outros', 'vitimas', 'vitimasfatais', 'velocidade_max_via'}

def tipo_logico(coluna):
    # o leitor carrega tudo como texto (object); aqui vai o tipo real do dado
    c = coluna.lower()
    if c == 'data':
        return 'data (AAAA-MM-DD)'
    if c == 'hora':
        return 'hora (HH:MM:SS)'
    if c in NUMERICAS:
        return 'inteiro (contagem; vírgula decimal a partir de 2022)'
    if c == 'protocolo':
        return 'identificador'
    return 'texto'

dados = []

for arquivo in res['arquivos']:
    df = ler_csv_robusto(arquivo)

    for coluna in df.columns:
        valores_validos = df[coluna].dropna()

        dados.append({
            'Arquivo': arquivo.name,
            'Variável': coluna,
            'Tipo': tipo_logico(coluna),
            'Descrição': descrever(coluna, df[coluna]),
            'Exemplo': (
                str(valores_validos.iloc[0])
                if not valores_validos.empty
                else 'Sem dados'
            ),
            '% Nulos': round(df[coluna].isna().mean() * 100, 2)
        })

dicionario = pd.DataFrame(dados)

display(dicionario)
```

```python
pendentes = dicionario[
    dicionario['Descrição'] == 'Descrição a validar'
]

for coluna in sorted(pendentes['Variável'].unique()):
    exemplos = (
        pendentes[pendentes['Variável'] == coluna]['Exemplo']
        .drop_duplicates()
        .head(2)
        .tolist()
    )
    print(f"{coluna} | Exemplos: {exemplos}")
```
